# <font color = "ADD8E6">***Manipulção de Variáveis e suas utilizadades***</font>

  Quando abordamos ***Variáveis*** na Análise de Dados, É necessário entendermos que existem tipos delas: **<font color = "FF8C69">Quantitativa</font> e <font color = "FFA07A">Qualitativa</font>**.

  - ## <font color = "FF8C69">**Qualitativas**</font> **(Categóricas)**:
    Se refere a variáveis não numéricas que definem grupos ou categorias:
    
    - **Nominais**: Não possui hierarquia ou ordem entre as categorias. (**Exemplo**: Sexo, Profissão, Estado civil).
    - **Ordinais**: Existe uma ordem ou hierariqeuia natural entre os dados. (**Exemplo**: Grau de escolaridade (fundamental, médio, superior)).

  - ## <font color = "FFA07A">**Quantitativas**</font> **(Númericas)**:
    Expressa contagens ou medições em números (sendo o contrário das *Qualitativas*):

    - **Discretas**: São resultados de contagens, e se expressam em valores inteiros. (**Exemplo**: Número de filhos, Quantidade de carros).
    - **Contínua**: São resultados de medições, e podem se expressarem na forma de qualquer valor em uma escala numérica, incluindo frações ou decimais (**Exemplos**: Altura, Idade, Sálario).

# **Tratamento De Dados**

### Durante o consumo de um banco de dados, um dos pontos mais importantes do processo, é tratar os tipos de dados que impedem que ele esteja limpo. Tais como:
  - **Valores nulos**: Dados inexistentes ou que não possuem informação em um banco de dados. determinam que algum campo, célula ou variável não possui um valor informado ou conhecido.
  - **Dados duplicados**: São Dados que possuem o mesmo valor em diferentes registros dentro do banco de dados. São prejudiciais pois impedem uma análise confiável das informações, sendo dificultador nas tomadas de decisão.
    - **Tipos de Duplicação**:
      - Duplicação de Linhas: Quando dois registros em uma tabela possuem os mesmo valores em todas ou na maioria das colunas, representando a mesma entidade ou evento mais de uma vez de forma desnecessária.
      - Duplicação de Colunas: Quando uma tablea armazena o mesmo tipo de informação em dois campos diferentes, ou quando uma consulta SQL retorna colunas com o mesmo nome.

## **Teste em Campo:**

````python
df.isnull().sum()
# Retorna os valores nulos por coluna, junto mostra a quantidade de valores nulos.
````

````python
l_duble = df.duplicated().sum() # Retorna linhas duplicadas, e a quantidade existentes dela

# Armazenei a quantidade de linhas duplicadas na minha variável "l_duble" -> lines_duplicated
print("Quantidade de linhas duplicadas:", l_duble)
````

### **Agora vou tratar esses dados para que o banco de dados se mantenha limpo:**

````python
# Vou utilizar a função drop_duplicates() para remover as linhas duplicadas

df_sem_duplicatas = df.drop_duplicates().copy() ## Criando nova variável para armazenar o banco de dados sem as linhas duplicadas


l_duble = df_sem_duplicatas.duplicated().sum()
print("Quantidade de linhas duplicadas:", l_duble)
````

````python
# Agora vou criar a tradução das colunas e algumas linhas:

colunas_pt_br = {
    "work_year": "ano_trabalho",
    "job_title": "cargo",
    "job_category": "categoria_cargo",
    "salary_currency": "moeda_salario",
    "salary": "salario",
    "salary_in_usd": "salario_dolar",
    "employee_residence": "residencia_funcionario",
    "experience_level": "nivel_experiencia",
    "employment_type": "tipo_contrato",
    "work_setting": "modelo_trabalho",
    "company_location": "localizacao_empresa",
    "company_size": "porte_empresa"
}

modeloTrabalho_pt_br = {
    "Remote": "Remoto",
    "Hybrid": "Híbrido",
    "In-person": "Presencial"
}

port_empresa_pt_br = {
    "S": "Pequeno",
    "M": "Médio",
    "L": "Grande"
}


tp_contrato_pt_br = {
    "Full-time": "Tempo Integral",
    "Contract": "Contrato",
    "Part-time": "Meio Período",
}

nivel_exp_pt_br = {
    "Mid-level": "Pleno",
    "Entry-level": "Júnior",
    "Senior": "Sênior",
    "Executive": "Executivo"
}

df_pt_br = df_sem_duplicatas.rename(columns=colunas_pt_br) # Renomeando as Colunas para Portugues
display(df_pt_br.columns) # Faz a visualização unicamente das colunas

df_pt_br['modelo_trabalho'] = df_pt_br['modelo_trabalho'].replace(modeloTrabalho_pt_br) # Altera os dados da linha de modelo de trabalho para PT-BR
df_pt_br['porte_empresa'] = df_pt_br['porte_empresa'].replace(port_empresa_pt_br) # Altera os dados da linha de porte da empresa para PT-BR
df_pt_br['tipo_contrato'] = df_pt_br['tipo_contrato'].replace(tp_contrato_pt_br) # Altera os dados da linha de tipo de contrato para PT-BR
df_pt_br['nivel_experiencia'] = df_pt_br['nivel_experiencia'].replace(nivel_exp_pt_br) # Altera os dados da linha do nível de experiência para PT-BR
````

# **PROJETO FINAL**


````python
df_analyst = df_pt_br[df_pt_br['cargo'] == 'Data Analyst']
````

````python
media_salario = (
    df_analyst
    .groupby(['ano_trabalho', 'nivel_experiencia'])['salario_dolar']
    .mean()
    .reset_index()
)
````

````python
pivot = media_salario.pivot(
    index='ano_trabalho',
    columns='nivel_experiencia',
    values='salario_dolar'
)

ordem = ['Júnior', 'Pleno', 'Sênior']
pivot = pivot.reindex(columns=ordem)
````

````python
plt.figure(figsize=(4,2))
pivot.plot(marker='o')

plt.title("Média salarial por Ano e Experiência (Analista de Dados)")
plt.xlabel("Ano")
plt.ylabel("Salário Médio (USD)")
plt.grid()

plt.legend(title='Experiência')

plt.show()
````

````python
profissionais_ano = (
    df_analyst
    .groupby('ano_trabalho')
    .size()
)
````

````python
crescimento = profissionais_ano.pct_change() * 100
print(crescimento)
````

#### Impacto da pandemia na Análise de Dados (Profissão)

Durante o período da pandemia, houve uma transição radical no modelo em que o mundo funcionava, atividades presenciais necessárias passaram a ser modernizadas ao modo virtual/remoto. Esse movimento acelerou de forma repentina a digitalização de processos em empresas, governos e na sociedade como um todo.

Com a migração para o ambiente digital, como e-commerce, trabalho remoto, serviços por aplicativo e plataformas online - houve uma explosão no volume de dados gerados diariamente. Esse cenário elevou a importância de profissionais capazes de coletar, tratar e transformar dados em informação estratégica.

O crescimento observado no período pós-pandemia indica que essa demanda não foi temporária. Pelo contrário, consolidou-se como uma tendência de longo prazo. Empresas passaram a depender ainda mais de análises preditivas, indicadores de performance (KPIs) e inteligência de negócios para tomada de decisão em um ambiente incerto e altamente competitivo.

Em síntese, a pandemia atuou como um catalisador para a valorização da área de dados, ampliando significativamente tanto a oferta quanto a demanda por profissionais qualificados, e consolidando o dado como um dos principais ativos estratégicos da economia moderna atual.
